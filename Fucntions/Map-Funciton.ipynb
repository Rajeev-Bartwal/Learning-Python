# Map Function

the map() function is used or applies a given function to all items of an iterable (or an input list) and returns a map aobject( an iterator). This is particularly useful for transforming data in a list comprehensively.

```python
number = [1,2,3,4,5,6,7,8]

def square(lst):
  sq = []
  for i in lst:
    sq.append(i**2)

  return sq

print(square(number))
```

```python
## using the lambda and Map() function.

# Syntax map(lambda func , iterable)

list(map(lambda x : x*x , number))

```

```python
# for cube

list(map(lambda x : x**3 , number))
```

```python
# map multiple iterables

nums1 = [1,2,3,4]
nums2 = [5,6,7,8]

#add number of multiple list
added =list(map(lambda x,y : x+y , nums1 , nums2))
print(added)
```

```python
# map to conv to list of string in int

numbers = ['1','2','3','4']

int_num = list(map(int , numbers))

print(int_num)
```

```python
words = ['apples' , 'banana' , 'cherry']

upr_words = list(map(str.upper , words))

print(upr_words)
```

```python
# map to dictionaries

people = [{'name' : 'rajeev' , 'age' : 32},
          {'name' : 'kanishka' , 'age' : 22},
          {'name' : 'sheetal' , 'age' : 54},
          {'name' : 'anshu' , 'age' : 12}]

names = list(map(lambda x : x.get('name') , people))
print(names)

ages = list(map(lambda x : x.get('age') , people))
print(ages)
```

## Conclusion

The map() function is a powerful tool for applying transformations to iterable datastructure. it can be used with regular function, lambda functions and even multiple iteraters , providing a versatile approach to data proccessing in python . by understanding and utilizing map() , you can write more efficient and readable code.
