```python
def even_odd(num):
  if num%2 == 0:
    print("The number is even")
  else:
    print("The number is Odd")


#callig a funciton
even_odd(10)
```

```python
## function with multiple parameters

def add(a,b):
  return a+b

c = add(3,5)
print(c)
```

```python
## default paramenters

def greet(name):
  print(f"hello {name} welcome to the paradise")

greet("rajeev")
```

```python
def greet(name = "Guest"):
  print(f"hello {name} welcome to the paradise")

greet()
greet("kanishka")
```

```python
# variable length parameter
# potional and keyword arguments


# potional arguments
def addition(*args):
  total = 0
  for i in args:
    total += i
  print(total)


addition(1,2,3,4,5,6)
```

```python
# keyword arguments

def key_args_func(**kwargs):
  for key , val in kwargs.items():
    print(f"{key} : {val}")

key_args_func(name = "rajeev" , age = 34, city = "Dehradun")
```

```python
def print_details(*args,**kwargs):
  for key , val in kwargs.items():
    print(f"keyword args : {key} : {val}")
 
  for val in args:
      print(f"positional args : {val}")


print_details(1,2,3,4,5,6, "bartwal" ,name = "rajeev" , age = 34, city = "Dehradun")
```

```python
# Return Statement 
# function willl retur some value after function will execute

def multi(a,b):
  return a*b

print(multi(2,5))

# multiple return value

def multi_return_val(name, *args):
  return args ,name

multi_return_val("rajeev",1,2,3,45,4)
```
