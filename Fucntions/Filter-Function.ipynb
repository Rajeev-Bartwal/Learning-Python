# Filter() function in Python

The filter() function constructs an iterator from elements of an iterable for which a function return true.

it is used to filter out items from a list(or other iterable) based on a condition

```python
# syntax Filter(func , iterable)

def even(num):
  if num%2 == 0:
    return True
  else: 
    return False

print(even(24))
```

```python
lst = [12,3,5,6,7,8,55,4,2,5,11]

evens = list(filter(lambda x : x%2==0 , lst))
print(evens)
```

```python
# filter with lambda

number = [23,4,5,6,7,84,2,1]

greater_then_five = list(filter(lambda x : x>5 , number))
print(greater_then_five)
```

```python
# Filter with lambda and multiple condition

number = [24,4,5,6,7,84,2,14,7,3,2,5,8,99]

even_greater_then_five = list(filter(lambda x : x%2 == 0 and x>5 , number))
print(even_greater_then_five)
```

```python
# Filter() on dictionary age > 25

people = [{'name' : 'rajeev' , 'age' : 32},
          {'name' : 'kanishka' , 'age' : 22},
          {'name' : 'sheetal' , 'age' : 54},
          {'name' : 'anshu' , 'age' : 12}]

older_people = list(filter(lambda x : x.get('age') > 25 , people))
print(older_people)
```
