### count every word occurance with a function

```python
def count_freq(str):
  words = str.split()
  word_count = {}
  for word in words:
    word = word.lower().strip('.,!?;:"\'')
    word_count[word] = word_count.get(word,0)+1

  return word_count


str = "Rajeev bartwal rajeev bartwal is a very good software engineer and knows very diff deff languages"
print(count_freq(str))
```

### Valid Email or not

```python
def valid_email(email):
  if '@'  in email:
    if '.com' in email:
      return True
  else:
    return False

if valid_email('bartwalrajeev3@gmail.com'):
  print("valid email")
else :
  print("not a valid email")
```
