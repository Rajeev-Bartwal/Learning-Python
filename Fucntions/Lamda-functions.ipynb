# Lambda Functions in Python

lambda functions are small anonymous fucntions using lambda keyword, They can have any number of arguments but only one expression . They are commonly used for short operations or as arguments to higher-order funtions.

## syntax

Lambda arguments : expression

```python
def add(a,b):
  return a+b

addition = lambda a,b : a+b

print(add(6,8))

print(addition(2,6))
```

```python
even = lambda x :  x%2 == 0 
print(even(18))
```

```python
addition = lambda x,y,z : x+y+z

print(addition(9,6,4))
```

```python
## very useful in map and filter

number = [1,2,3,4,5,6,7]

# map funtion it applies a funciton to all ele. in a list or a datatype


list(map(lambda x : x**2 , number))
```
