## Tuples

Tuples are ordered , immutable collection of items. created with the (). They are similar to list, but immutability makes them diffrent.

```python
my_tuple = (1,2,3,4)
print(my_tuple)

# Way of creating the empty tuples
# 1.
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))

# 2.
emp_tup = tuple()
print(emp_tup)
print(type(emp_tup))
```

```python
# List -> tuple
nums = tuple([1,2,3,4,5])
print(nums)

# tuple -> List
numbers = list((2,4,5,6,7,8))

print(numbers)
```

```python
mixed_tup = (1, "hello" , True, 6.7)
print(mixed_tup)

print(mixed_tup[0])
print(mixed_tup[-1])
print(mixed_tup[1:3])
```

```python
## Tuple operation

concat_tup = nums + mixed_tup
print(concat_tup)
```

```python
mixed_tup * 3  #entire tuple will get appended 3 times
```

```python
# Immutablity of tuples
# Tuples are immutable you can not change their elements once assigned

my_tuple[1] = 3  #you will get error in this operation. you can not change the values of tuples
print(my_tuple)
```

```python
#Tuple Methods

print(nums.count(2))  #count the given elements ocurrence

print(nums.index(4))
```

```python
#packing and unoacking tuple

#packing
packed_tuple = 1,"hello" , 3.14   #will get autmatically packed in tuple
packed_tuple
```

```python
#unpacking

a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
# unpacking with *

# * will store multiple ele. of tuple in a variable as a list
numbers = (1,2,3,4,5,6)

first,middle,*last= numbers
print(first)
print(middle)
print(last)
```

```python
## nested tuple
# Nested List

list = [[1,2,3,4],(6,7,8,9),["hello",3.12,34]]
print(list[0][:3])
print(list[1][:3])

nested = (1,2,3,(4,5,6),[2,4,6,7])
nested[4][2] = 34
nested
```

```python
# iterating over nested tuple

for sub_tuple in nested:
  print(sub_tuple)
```
