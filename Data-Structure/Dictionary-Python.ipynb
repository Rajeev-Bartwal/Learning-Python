### Dictionaries

is a unorderd collection of items,They store data in key-value pairs. Keys must be uniwue and immutable(eg. strings numbers or tuple) , while values can be any type

```python
dicti = {'a': 1 , 'b' : 2, 'c' : 3}
print(dicti)

# creating an empty dictionary

d = {}
print(type(d))

#or

empty_dictionary = dict()
print(empty_dictionary)
```

```python
student = {"name" :"rajeev",
           "grade" : 32,
           "age" : 32}

print(student)
print(type(student))
```

```python
# Unique Key
student = {"name" :"rajeev", "grade" : 32,"name" :34}
print(student)

#name is a key so if you again put the value on same key so the previous value will be overrided
```

```python
# Accessing Dictionary Values

student = {"name" :"rajeev","grade" : 32,"age" : 32}

print(student['grade'])
print(student['age'])

# Accessing via default get() method
print(student.get('name'))
print(student.get('last-name'))
print(student.get('last-name' , "Not Avialable")) 
```

```python
print(student)
```

```python
student["age"] = 30 #update key value
print(student)

student["address"] = "Dehradun" # added a new key and value
print(student)

#removing a key and value
# student.pop("grade") 
# print(student)

#or using the del keyword
del student["age"]

print(student)
```

```python
# Dictionary Methods

keys = student.keys()  # All keys of dictionry
print(keys)

values = student.values()   # All Values of dictionry
print(values)

items = student.items()  # All Pairs of dictionry
print(items)
```

```python
# Shallow Copy

student_copy = student # this is a deep copy mena syou are directly assigning the ref of the original one to the copy

print(student_copy)
print(student)
```

```python
# This will change the value of name in both original and the copy

student['name'] = "sheetal"

print(student_copy)
print(student)


# How to make the shallow copy
stud_copy = student.copy()

#this will not be changing the name value in the stud_copy cause it is a shallow copy

student['name'] = "Rajeev"

print(stud_copy)
print(student)

```

```python
# Itrating Over Dictionaries

# we can use loops to iterate over dictionaries, viw keys values and items


# iterate over keys

for key in student.keys():
  print(key)
```

```python
# iterate over values

for value in student.values():
  print(value)
```

```python
# iterate over piars

for key, value in student.items():
  print(f"{key} :- {value}")
```

### Nested Dictionaries

```python
students = { 
   "student1" : {'name': 'Rajeev' , 'age' : 21},
   "student2" : {'name': 'Sheetal' , 'age' : 24}
}

print(students)
```

```python
## Access Nested Dictionary ele.

print(students['student1']['name'])
print(students['student1']['name'] , students['student1']['age'])
```

```python
## Iterating Over nested dictionaries

for student_id , student_info in students.items() :
  print(f"{student_id} : {student_info}")
  for key,val in student_info.items():
    print(f"{key} : {val}")
```

### Dictionary Comprehension


```python
# Using for loop make a dictionary of num and their squares 
dict = {}
for i in range(1,5):
  dict[i] = i**2

print(dict)

# Using comprehension make a dictionary of num and their squares 

sqares = {x:x**2 for x in range(1,5)}
print(sqares)




```

```python
# Using for-loop make a dictionary of even num and their squares 

even_sq_dict = {}

for x in range(1,10):
  if x%2 == 0:
    even_sq_dict[x] = x**2

print(even_sq_dict)
# Using coditional - comprehension make a dictionary of even num and their squares 

even_sq = {x : x**2 for x in range(1,10) if x%2 == 0}
print(even_sq)
 
```

```python
# use a dict to count the freq of chars in string

str = "rajeev bartwal"
result = {}

for ch in str:
  if ch != " ":
    result[ch] = result.get(ch , 0)+1

print(result)
```
