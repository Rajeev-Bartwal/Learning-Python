### Slicing List

```python
nums = [2,5,12,7,9,3,1]

#list[start-ind : end-ind : steps] start included , end excluded, steps how you want to iterate (jums step to ahead)
print(nums[1 : ])
print(nums[2 : 5])
print(nums[: 4])
print(nums[2: ])
print(nums[ :: 2])
print(nums[ :: -1])
```

### Iterating Over List

### List Comprehension

Basic syntax [expression for item in iterable]

with conditional logic [expression for item in iterable if condition]

Nested List Comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
# Basic List Comprehension 

# Square of all numbers

squares = [x**2 for x in range(10)]
print(squares)
```

```python
## List Comprehension with condition


# without comprehension

list = []
for x in range(10) :
  if x%2 == 0 :
    list.append(x**2)

print(list)
# sqaures of only even
#
even_squares = [x**2 for x in range(10) if x%2 == 0] 
print(even_squares)
```

```python
## Nested List Comprehension (like nested loop)

list1 = [1,2,3,4,5]
list2 = ['a', 'b' , 'c' , 'd' , 'e']

pair = [[i,j] for i in list1 for j in list2]

print(pair)
```

```python
# List Comprehension with fuction calls

words = ["rajeev" , "bartwal" , "rishab" , "mohit" , "negi"]

length = [len(word) for word in words]
print(length)
```

```python


result = ["even" if n%2 == 0 else "odd" for n in range(1,  8)]
print(result)
```
