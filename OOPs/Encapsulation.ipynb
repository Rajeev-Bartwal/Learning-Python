## Encapsulation and Abstraction

Encapsulation and Abstraction are two fundamental principle of object oriented programming (OOP) that help in designing robust, maintainable, and reusbale code. Encapsulation involves bundling data and methods that operate on data with in a single unit, while abstraction invloves hiding complesx implementation details and exposing only the necessary features.


## Encapsulation

It is the concept of binding or wrapping data ( variables) and methods(func) together as a single unit. It restricts direct access to some of the objects components which is a means of preventing accidental interference and misuse of the data.

```python
# We can do encapsulation by using Getter and setter Methods

# public , private and private access modifires

# Encapsulation

class Person:
  def __init__(self , name , age):
    self.name = name  # public variable
    self.age = age     # public variable

person = Person('rajeev' , 23)
print(person.name)

dir(person)
```

```python
# __ by addig __ two underscores in front of the variable name in init method we can make them private.

class Person:
  def __init__(self , name , age):
    self.__name = name  # private variable
    self.__age = age    # private variable
  
  
person = Person('rajeev' , 23)

dir(person)
```

```python
# _ by addig _ single underscores in front of the variable name in init method we can make them Protected.

class Person:
  def __init__(self , name , age):
    self._name = name  # Protected variable
    self._age = age    # Protected variable
  
class Employee(Person):
  def __init__(self , name , age):
    super().__init__(name , age)

emp = Employee('rajeev' , 23)
print(emp._name)
```

```python
# Getter and Setter Encpsulation

class Person:
  def __init__(self , name , age):
    self.__name = name  # Private variable
    self.__age = age    # Private variable

  # getter method for name
  def getName(self):
    return self.__name

  # setter for name
  def setName(self, name):
    self.__name = name

  def getAge(self):
    return self.__age

  # setter for age
  def setAge(self, age):
    if age > 0:
      self.__age = age
    else :
      print('Age can not be negative')


p = Person('rajeev' , 23)

print(p.getName())
print(p.getAge())

p.setName('kanishka')
p.setAge(-9)

p.getAge()

p.setAge(24)

print('somthing')


print(p.getAge())
print(p.getName())
```
