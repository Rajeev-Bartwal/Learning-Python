## Operator Overloading

Operator Overloading allows you to define the behaviou of operators (+,_,*,%) for coutom objects.You achive this by overriding the specific magic methods in your class.

```python
### Some of the operator overloading dunder methods

'''
__add__(self, other) : Adds two objects using the + operator
__sub__(self , other) : Subbtracts two objects using the - operator
___mul__(self , other) : Multiplies 2 objects using the * operator
__truediv__(self,other) : Divides 2 objects using the / operator
__lt___(self , other) : Checks if one object is less than another using the < operator
__eq__(self , other) : Checks if one object is equal to the another using the == operator

__gt__ : greater then
'''
```

```python
### Mathematical operation for vectors.

class Vector:

  def __init__(self , x , y):
    self.x = x
    self.y = y

  def __add__(self , other):
    return Vector(self.x + other.x , self.y + other.y)

  def __sub__(self , other):
      return Vector(self.x - other.x , self.y - other.y)

  def __eq__(self , other):
      return Vector(self.x == other.x , self.y == other.y) 

  def __repr__(self):
    return f"Vector( {self.x} , {self.y})"


v1 = Vector( 3, 6)
v2 = Vector( 4, 8)

print(v1+v2)
print(v2-v1)
print(v1 == v2)
```

```python
class Complex:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    # + operator
    def __add__(self, other):
        return Complex(
            self.real + other.real,
            self.imag + other.imag
        )

    # - operator
    def __sub__(self, other):
        return Complex(
            self.real - other.real,
            self.imag - other.imag
        )

    # * operator
    def __mul__(self, other):
        real = (self.real * other.real) - (self.imag * other.imag)
        imag = (self.real * other.imag) + (self.imag * other.real)

        return Complex(real, imag)

    # / operator
    def __truediv__(self, other):
        denominator = (other.real ** 2) + (other.imag ** 2)

        real = ((self.real * other.real) +
                (self.imag * other.imag)) / denominator

        imag = ((self.imag * other.real) -
                (self.real * other.imag)) / denominator

        return Complex(real, imag)

    # == operator
    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag

    # Display complex number
    def display(self):
        print(f"{self.real} + {self.imag}i")


# Creating objects
c1 = Complex(3, 4)
c2 = Complex(5, 2)

# Addition
c3 = c1 + c2
print("Addition:")
c3.display()

# Subtraction
c4 = c1 - c2
print("Subtraction:")
c4.display()

# Multiplication
c5 = c1 * c2
print("Multiplication:")
c5.display()

# Division
c6 = c1 / c2
print("Division:")
c6.display()

# Equality
print("Are c1 and c2 equal?", c1 == c2)
```
