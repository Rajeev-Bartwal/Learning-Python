## Classes and Objects

Object-Oriented-Programming ( OOP ) is a programming paradigm that uses 'objects' to design application and the computer programs.
OOP allow for modeling real world scenarios using classes and objects. This lesso cover the basic of creating classes and objects, including instance valiables and methods.

```python
# A class is a blue print for creating objects. can have Attributes, Mehtods

class person:
  
  #instance variables and methods


  # Constructor or initializer of class (not a constructor)
  # self is a keyword means point to current object of a class
  def __init__(self,name,age):
    self.name = name
    self.age = age

# this is a method of a class and you can access the instance var. by self self means the current obj who is calling this method.
  def walk(self):
    print(f'{self.name} is waliking')

a = person("rajeev", 23)
print(a.name)
print(a.age)
a.walk()


b = person("sheetal", 24)
print(b.name)
print(b.age)
b.walk()

c = person("kanishka",23)

print(c.name)
print(c.age)
c.walk()
```

```python
## Modeling a bank acount

class BankAccount:

  def __init__(self,owner , balance=0):
    self.owner = owner
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount
    print(f'account has been creadited {amount} and balance is {self.balance}')

  def withdrawl(self , amount):
    if self.balance < amount:
      print('Insufficient Balance !')
    else:
      self.balance -= amount
      print(f'Account has been deposited {amount} hand Balance : {amount}')

  def get_balance(self):
    print(f'Account Balance is : {self.balance}')


# Creating Account.
Rajeev_acc = BankAccount('Rajeev Bartwal')

# Getting Balance.

Rajeev_acc.get_balance()

# Adding money
Rajeev_acc.deposit(10000)

# again getting balance
Rajeev_acc.get_balance()


# withdrawling insufficient balance
Rajeev_acc.withdrawl(12000)

# againg withrawling
Rajeev_acc.withdrawl(5000)

# agian adding money to bank.
Rajeev_acc.deposit(10000)

```
