# Inheritance in Python

inheritance is a fundamental pillar r concept of OOPs that allows a class to inherit attributes and methods/properties from another parent class. This lesson covers single inheritance and multiple inheritance demonstrating how to create and use them in python

```python
# Single inheritance

# parent class
class Car:

  def __init__(self, windows , door , enginetype):
    self.door = door
    self.windows = windows
    self.enginetype = enginetype

  def drive(self):
    print(f'The person will drive the {self.enginetype} car')


    
```

```python
car1 = Car(4,5,'petrol')
car1.drive()
```

```python
# child class inheriting the Car class properties.

class Tesla(Car):

  def __init__(self, windows, door, enginetype , is_Self_Driving):
    super().__init__(windows, door, enginetype)
    self.is_Self_Driving = is_Self_Driving

  def self_Driving(self):
    print(f'Tesla supports self driving : {self.is_Self_Driving}')
```

```python
tesla1 = Tesla(4 ,4 , 'Electric' , True)

tesla1.self_Driving()

tesla1.drive()
```

```python
## Multiple Inheritance.

# When a class inherits from more then one class.


# Base class 1
class Animal:

  def __init__(self , name):
    self.name = name

  def speak(self):
    print('subclass must implement this method')


  
```

```python
# Base class 2

class Pet:
  def __init__(self , owner):
    self.owner = owner

  
```

```python
## Derived class
class Dog(Animal , Pet):

  def __init__(self , name , owner):
    Animal.__init__(self, name)
    Pet.__init__(self , owner)

  def speak(self):
    return f"{self.name} say woof"

alice =  Dog('alice' , 'Rajeev')

alice.speak()
```
