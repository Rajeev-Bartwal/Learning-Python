## Abstraction

Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object to the user. This helps in reducing programming complexity and effort.

```python
## Abstraction ABC's

from abc import ABC , abstractmethod

# Abstract class
class Vehical(ABC):

  def drive(self):
    print('the vehical is used for driving')

  @abstractmethod
  def strat_engine(self):
    pass

class Car(Vehical):

  def strat_engine(self):
    print('car engine is started')

def operate_vehical(Vehical):
    Vehical.strat_engine()
    Vehical.drive()

car = Car()

operate_vehical(car)
```
