## Generators
( Generators is subclass of Iterators )

Generators are as simple way to create iterators. They use yield keyword to produce a series of values lazily ( means when value is needed it will ad it in to iterator), Which means they generate values on the fly and do not store them in memory.

```python
def square(n):
  for i in range(3):
    yield i**2
```

```python
square(3)
```

```python
for i in square(3):
  print(i)
```

```python
a = square(3)
```

```python
next(a)
```

```python
def my_generator():
  yield 1
  yield 2
  yield 3
```

```python
gen = my_generator()
```

```python
print(next(gen))
```

```python
for val in gen:
  print(val)
```

### Practical example : Reading Large Files
Generators are particularly useful for reading large files beacuse they allow you to process one line at a time without loading the entire file into memory

```python
## Practical Example of generators :
# Reading Large files

def read_large_file(file_path):
  with open(file_path , mode='r') as file :
      for line in file:
        yield line
```

```python
row = read_large_file('large_file.txt')
```

```python
print(next(row))
```

```python
for line in read_large_file('large_file.txt'):
  print(line.strip())
```
