### Closure in Python:
A closure is a function that remembers and can access variables from its outer function even after the outer function has finished executing.

Simple: “Inner function apne outer function ke variables ko yaad rakhta hai.”

```python
# Closures functions - Mehoitd 

def outer(msg):
  
  def inner():
    print("Welcom to the learning")
    print(msg)
    print("Hello hi")

  return inner()
```

```python
outer("welcome")
```

```python
def multiplier(n):  # outer method
    def multiply(x): # Inner Method
        return x * n
    return multiply # Returning the inner method

double = multiplier(2) 
triple = multiplier(3)

print(double(5))   # 10  #inner will remeber 2 means n var of outter
print(triple(5))   # 15

# double remembers n = 2, while triple remembers n = 3.
```

```python
# Decorator


def outer(func):
  
  def inner():
    print("Welcom to the learning")
    func()
    print("Hello hi")

  return inner()
```

```python
def cource_into():
  print("This is an advance pytho course")
```

```python
outer(cource_into)
```

```python
@outer
def cource_into():
  print("This is an advance pytho course")

```

```python
def decorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper


```

```python
@decorator
def hello():
    print("Hello")

hello()
```

```python
## Decorator With Arguments 
# Decorator mein wrapper return karo, execute mat karo; aur arguments ko flexible rakhne ke liye *args, **kwargs use karo.

def my_deco(func):
  def wrapper(*args , **kwargs):
    print("somthing is happening before the function is called")
    func(*args , **kwargs)
    print("somthing is happening after the function is called")
  return wrapper
```

```python
@my_deco
def square(num):
  lst = []
  for i in range(num):
    lst.append(i**2)
  print(lst) 

square(5)
```
