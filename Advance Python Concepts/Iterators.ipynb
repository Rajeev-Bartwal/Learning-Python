## Iterators

Iterators are advance python concept that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequetially without exposing the underlying structure.

```python
my_list = [1,2,3,4,5]

for i in my_list:
  print(i)

print(type(my_list))
print(my_list)
```

```python
# Iterator is created with iter() it will make the collection a iterator 

iterator = iter(my_list)

print(type(iterator))
```

```python
# Next() in iterator 
# It uses a Lazy loading functionality (means it will not give us any element unless we want to access it or we iterate through that element)

# Iterate through all the elements we will use next()

print(next(iterator))
```

```python

try:
  print(next(iterator))
except StopIteration:
  print('There are no elemets in the iterator')
```
