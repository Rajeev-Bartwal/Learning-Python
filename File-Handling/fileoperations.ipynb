## File operations - Read and Write Files

file handling is a crucial part of programming language. Python provdes built in functions and methods to read from and write to files.
both text and binary . this lesson will cover the basics of file handling including reading and writing text files and binary files.


```python
### read a whole file

with open('example.txt' , mode='r') as file :
  content = file.read()
  print(content)
  

```

```python
## Read a file line by line

with open('example.txt' , mode='r') as file :
  for line in file:
    print(line.strip())  ##.strip() operation removes the new line character
```

```python
## opening a file in write mode will remove the previous content on the write operation. so be carefull with ('w'  write mode)
with open('example.txt' , mode='w') as file :
  file.write('Hello world\n')
  file.write('this is a new content\n')
```

```python
# write a file without overwritting with append mode use 'a'

with open('example.txt' , mode='a') as file :
  file.write('With append mode operation\n')
  file.write('the old content will be not overriden\n')
```

```python
# Wirte multiple lines or list or lines


lines = ["first lin\n" , "sec. line\n" , "third line\n"]
with open('example.txt' , mode='a') as file :
  for line in lines:
    file.write(line)
```

```python
## or using a funtion file.writelines()

lines = ["first lin\n" , "sec. line\n" , "third line\n"]
with open('example.txt' , mode='a') as file :
  file.writelines(lines)
```

```python
## Working with binary file


# Write binary 
data = b'\x00\x02\x03\x04'
with open('example.bin' , mode='wb') as file:
  file.write(data)
```

```python
# Read binary

with open('example.bin' , mode='rb') as file:
  content = file.read()
  print(content)
```

```python
## REad the content from a source text file and write to a destination txt file

with open('example.txt' , mode='r') as source:
  content = source.read()
  print(content)

with open('destination.txt' , mode='w') as destination:
  destination.write(content)
```

```python
## Read a text file and count the no of lines , words and chars
# 
def count(file_path):

  with open(file_path ,mode='r') as file:
  
    lines = file.readlines()
    line_count = len(lines)
    word_count = sum(len(line.split()) for line in lines) 
    char_count = sum(len(line) for line in lines)

  return line_count , word_count , char_count

lines , words , chars = count('example.txt')
print(f'line : {lines} , words :  {words} , characters : {chars}')
```
