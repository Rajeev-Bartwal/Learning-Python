# Understanding Exceptins

Exception handling in python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program. this will cover the basics of exception, including how to use try, except ,else and finally block.

## What are exceptions ?

Exceptions means the events tht distrup the normal flow of the program. They occure when an error is encounterd during program execution. Common exceptions include:

1. ZeroDivisionError : Dividing by zero
2. FileNotFoundError : file not found
3. ValueError : Invalid value
4. TypeError : invalid type

```python
a , b = 10 , 0
try : 
  a/b 
  # c=d
except NameError as e:
  print('name' , e)
except Exception as e:
  print('zero' , e)
finally :
  print("This block will always execute")

```

```python
try:
  num = int(input("Enter a number"))
  result =  10/num
except ValueError:
  print("This is not a valid number")
except ZeroDivisionError as ex:
  print("Exnter a valid num in denominator gretaer than zero")
except Exception as ex:
  print(ex)
```

```python
## Try except  and else block
# whenerver a exception will come the else will not getting executed but if there is no exception then the else will get executed after try.

try:
  num = int(input("Enter a number"))
  result =  10/num
except ValueError:
  print("This is not a valid number")
except ZeroDivisionError as ex:
  print("Exnter a valid num in denominator gretaer than zero")
except Exception as ex:
  print(ex)
else :
  print(result)
```

```python
## Try , except , else abd finally block
# Finally will getting executed anyway if there is an exception or not .
# it will be getting executed used for resource closing.

try:
  num = int(input("Enter a number"))
  result =  10/num
except ValueError:
  print("This is not a valid number")
except ZeroDivisionError as ex:
  print("Exnter a valid num in denominator gretaer than zero")
except Exception as ex:
  print(ex)
else :
  print(f'This is the result of try block : {result}')
finally:
  print("Program ended")
```

```python
# Finally will getting executed even if you return somthign from else or try.

def input_taking():

  try:
    num = int(input("Enter a number"))
    result =  10/num
  except ValueError:
    print("This is not a valid number")
  except ZeroDivisionError as ex:
    print("Exnter a valid num in denominator gretaer than zero")
  except Exception as ex:
    print(ex)
  else :
    print(f'This is the result of try block : {result}')
    return result
  finally:
    print("Program ended")


input_taking()
```

```python
## File Handling and Exception Handling

try:
  with open('example1.txt' , mode='r') as file:
    content = file.read()
    
    print(content)
except FileNotFoundError:
  print("The file does not exist")
except Exception as e:
  print(e)

finally :
  if 'file' in locals() or not file.closed():
    file.close()
    print('file is closed')

```
