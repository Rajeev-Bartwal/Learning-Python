# Sandard Library Overview

Pythons standard library is a vast collection of modules and packages that come bundled with python, providing a wide range of functions out of the box. Here's an overview of aome of the modt commonly used modules and packages in the python standard library.

```python
import array

arr = array.array('i' , [1,2,3,4,5,6])
print(arr)
```

```python
import math

print(math.sqrt(81))
print(math.pi)
```

```python
# random give us a random number
import random

print(int(random.randint(1,100)))

# will return any random item from the iterable or a sequence
print(random.choice(['apple' , 'banana' , 'cherry' , 'guava']))
```

```python
# os
# For File and Directory Access

import os

print(os.getcwd())
```

```python
os.makedirs('test_dir')
```

```python
# High level operations on files and collections of files

import shutil

shutil.copyfile('source.txt' , 'destination.txt')
```

```python
## Data serialization

# cov. dictiornaries  in JSON

import json

data = {'name' : 'rajeev','age' : 23}

json_data = json.dumps(data)
print(json_data)
print(type(json_data))

parsed_data = json.loads(json_data)
print(parsed_data)
print(type(parsed_data))
```

```python
## CSV Files

import csv


with open('example.csv' , mode='w' , newline='') as file:
  writer = csv.writer(file)
  writer.writerow(['name' , 'age'])
  writer.writerow(['rajeev' , 23])

with open('example.csv' , mode='r') as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
```

```python
# Date and time

from datetime import datetime,timedelta

# we can get curren time and date form datetime
print(datetime.now())


# we can manipulate or get previous or next date with timedelta
yesterday = datetime.now() - timedelta(days=1)
print(yesterday)
```

```python
# time

import time
print(time.time())

# program will sleep for 2 sec.
time.sleep(2)

print(time.time())
```

```python
# regular expression

import re

pattern = r'\d+'  # digits

text = 'there are 123 apples'

match = re.search(pattern , text)

print(match.group())
```
