## Import Modules in Python : Modules and Packages

In Python, modules and packages helps organize and reuse code .
Here's a comprehensive guide on how to import them.


```python
import math

print(math.sqrt(81))
```

```python
from math import sqrt , pi

print(sqrt(16))
print(pi)
```

```python
import numpy as np

nums = [1,4,2,6,7,4]

np.array(nums)

```

```python
from math import *
print(sqrt(81))
print(pi)
```

```python
from demo.maths import addition,multi

print(addition(3,6))

print(multi(4,7))
```
